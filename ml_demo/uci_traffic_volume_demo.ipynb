- Dataset: `Metro Interstate Traffic Volume` от UCI, 48,204 почасови наблюдения

- Regression модел: `Ridge` за предсказване на `traffic_volume`
- Classification модел: `LogisticRegression` за `low / medium / high` traffic bands
- Split: chronological `70/15/15`
- Test regression result: `R² = 0.7162`, `RMSE = 1056.86`
- Test classification result: `Accuracy = 0.7960`, `F1 macro = 0.7951`
