# Tabular ML Demo: Congestion Classification

Този notebook описва третия ML модел в проекта: `RandomForestClassifier` за congestion prediction върху `sofia_route_network.csv`.

## Dataset

- Файл: `../sofia_route_network.csv`
- Тип: tabular dataset
- Използвани feature-и: speed, lanes, congestion fields, traffic profiles, green profiles, road class, direction
- Таргет: `congestion band` (`low / medium / high`) за multi-hour сценарий (`hour=0..23`)

## Модел

- `RandomForestClassifier`
- Обучение върху репрезентативна извадка от dataset-а

## Резултат

- Accuracy: 0.8433
- Precision macro: 0.8459
- Recall macro: 0.8433
- F1 macro: 0.8440
