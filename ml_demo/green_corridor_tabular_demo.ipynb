# Tabular ML Demo: Green Corridor Classification

Този notebook е кратка notebook-форма на отделния демонстрационен ML проект в репото. Задачата е **tabular binary classification** върху routing dataset-а на София.

## Dataset

- Файл: `../sofia_route_network.csv`
- Размер: 91,686 rows
- Тип задача: таблични данни
- Таргет: `green_corridor`

## Модел

- NumPy logistic regression classifier
- Train/test split: 80/20
- Feature set: numerical road features, one-hot categorical features, и профилни масиви

## Резултат

- Accuracy: 0.9993
- Precision: 1.0000
- Recall: 0.9662
- F1 score: 0.9828
